**Methods to create numpyarray**

```python
# zeros()
# ones()
# full()
# arange()
# identity()
# eye()
```

```python
# np.zeros() -> it is used to create an array full of 0
arr = np.zeros(6,dtype = int)
arr
```

```python
# np.zeros() -> it is used to create an array full of 0
arr = np.zeros((4,3),dtype = int)
arr
```

```python
# np.ones() -> used to create an array full of 1
arr = np.ones((4,3),dtype = int)
arr
```

```python
# np.full() -> shape,value, fill the entire array with value
arr = np.full((4,3),100)
arr
```

```python
# create an array in the given range
arr = np.arange(10)
arr
```

```python
arr = np.arange(2,11,3)
arr
```

```python
for i in range(2,11,3):
  print(i,end = " ")
```

```python
# identity() -> creates a identity matrix
arr = np.identity(5,dtype = int)
arr
```

```python
# np.identity => rows == coloumn
# np.eye => rows != coloumn
arr = np.eye(4,3,dtype = int)
arr
```

**Arithmetic Operation in Numpy**

```python
l1 = [1,2,3,4,5]
l2 = [6,7,8,9,10]
l1 + l2
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
arr1 + arr2
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
arr1 - arr2
```

```python
# mean -> average -> sum of all the elements/number of elements
arr = np.array([10,20,21,25,35,45])
arr
```

```python
ans = np.mean(arr)
print(ans)
```

```python
# median -> middle value
# median -> odd length -> 1 median
# even length -> 2 median => average of 2 median
arr = np.array([10,20,21,25,35,45,50])
ans = np.median(arr)
print(ans)
```

```python
#sort
arr = np.array([11,1,15,4,21,13])
arr_sorted = np.sort(arr)
arr_sorted
```

```python
#descending order
arr = np.array([11,1,15,4,21,13])
arr_sorted = np.sort(arr)[::-1]
arr_sorted
```

```python
import numpy as np
arr = np.array([10, 50, 5, 80, 25])
-10 -50-5-,25
3,1,4,.......
# Method 1: Negate the array before argsort
desc_idx = (-arr).argsort()
sorted_desc = arr[desc_idx]
print("Descending indices:", desc_idx) # [3 1 4 0 2]
print("Sorted descending:", sorted_desc) # [80 50 25 10 5]
```
