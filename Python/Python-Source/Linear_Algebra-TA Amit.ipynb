```python
import numpy as np
a1 = np.array([[1,2,3],[5,6,7]])
print(a1)
```

```python
#Operations On the matrix

m1 = np.array([[30,40,50],[10,15,30]])
m2 = np.array([[5,6,7],[8,9,10]])
print(m1)
```

```python
print(m2)
```

```python
#Addition->
#RULE -> Two matrics can be add iff they are having the same size

print(m1+m2)
```

```python
print(m1-m2)
```

```python
#Scaler Multiplication -> Multiplying a matrix by a single number
m2 = np.array([[100,200,300],[10,30,40]])
print(m2)
```

```python
print(m2*2)
```

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5],[6]])
```
