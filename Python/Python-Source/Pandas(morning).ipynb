**Pandas **

It is a python library which is used for Data Manipulation and Data Analysis. It helps you to load, clean, Transform, analyze and explore the data. It stands for Pannel Data.

**Why was pandas created **
As we know Numpy is amazing for working with numbers, arrays and perform mathematical operations. But real - world data is messy :- missing values, mixed data, it is not clean,CSV/Excel file so, pandas was created to handle the real world problems, not only just for mathematics

**Pandas helps in -**

```python
# 1 - Load the dataset(CSV, Excel........)
# 2 - To clean the data(remove the null values, duplicate values, to handle null values,wrong values)
# 3 - to maipulate the data(filter, sort, group, merge...)
# 4 - to analyze the data for Machine Learning Models
```

**Series**
    it is one-dimensional array labeled array
    it is simply one column data with an index
    it can store number, string or mixed datatype

**Series = Data + Index**

```python
# Creating a Series -
```

**1- From a list**

```python
# to create a series ----> pd.Series
s = pd.Series([30,40,50,60,61,80,90])
print(s)
```

```python
print(type(s))
```

**2 - From a python dictionary**

```python
d1 = {'A':101, 'hii':102, 103:'Hello', 'A':'red'}
s2 = pd.Series(d1)
print(s2)
# Note - Keys in dictionary act as label
# values in dict act as data in series
```

**3 - from tuple**

```python
t = (31,32,33,34,46)
s4 = pd.Series(t)
print(s4)
```

```python
s = pd.Series([10,20,[30,40],50])
print(s)
print(type(s))
```

**4 - Custom Label/index**

```python
a1 = pd.Series([100,200,300,400], index = ['a','b','c','d'])
print(a1)
```

**Two more important methods for accessing value from data**
**1- .loc -> Label Based Access**

*   it is used to access data using index labels explicitly



```python
x1 = pd.Series([100,200,300,400,500], index = ['a','b','c','d','e'])
print(x1)
```

```python
# single values
x1.loc['d']
```

```python
# multiple label values
x1.loc[['a','b','d']]
```

```python
# label slicing
x1.loc['b':'d']
```

**2- .iloc -> integer position - position**

```python
x1 = pd.Series([100,200,300,400,500], index = ['a','b','c','d','e'])
print(x1)
```

```python
x1.iloc[3]
```

```python
x1.iloc[[2,3,4]]
```
