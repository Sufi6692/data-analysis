**Pandas **

**Why was pandas created **
As we know Numpy is amazing for working with numbers, arrays and perform mathematical operations. But real - world data is messy :- missing values, mixed data, it is not clean,CSV/Excel file so, pandas was created to handle the real world problems, not only just for mathematics

**Pandas helps in -**

```python
# 1 - Load the dataset(CSV, Excel........)
# 2 - To clean the data(remove the null values, duplicate values, to handle null values,wrong values)
# 3 - to maipulate the data(filter, sort, group, merge...)
# 4 - to analyze the data for Machine Learning Models
```

```python
# groupby() ---> divide the data into groups
```

```python
x2 = pd.DataFrame({'city':["Banglore","Delhi","Banglore","Mumbai","Banglore"],"Sales":[30,40,50,60,70]})
print(x2)
```

```python
result = x2.groupby('city', as_index=False).agg(
    Average_Sales=('Sales', 'mean')
)

print(result)
```

```python
# avg sales based on city
```

```python
x2.groupby('city')['Sales'].mean()
```

**Handling the null or missing values**

```python
import numpy as np
```

```python
df2 = pd.DataFrame({'age':[10,20,np.nan],'gender':['m','f','f'],'name':['riya',np.nan,np.nan]})
```

```python
df2
```

```python
df2.isnull()
```

```python
df2.isnull().sum() # give the count of null values in each column
# df2.count()
```

```python
df2.isnull().sum().sum()
```

```python
# How to handle them??
# 1 - Delete the null values ---> df.dropna(inplace=True)
# 2 - fill the null values with other values(mean,median,mode,constant value) --> df.fillna(value,inplace=True)
```

```python
df2
```

```python
df2['age'].fillna(0,inplace=True)
```

```python
df2
```

```python
df2.dropna(inplace = True)  # remove the entire row where null value is present
```

```python
df2
```

```python
df2.isnull().sum()
```
